```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('country_wise_latest.csv')
print(df.head())
print(df.describe())
```

```python
mean_value = df['Confirmed'].mean()
median_value = df['Confirmed'].median()
mode_value = df['Confirmed'].mode()[0]
print('Mean:', mean_value)
print('Median:', median_value)
print('Mode:', mode_value)

```

```python
confirmed_array = df['Confirmed'].to_numpy()
print('NumPy Mean:', np.mean(confirmed_array))
print('NumPy Median:', np.median(confirmed_array))
print('SciPy Mode:', stats.mode(confirmed_array, keepdims=True))
```

```python
group_summary = df.groupby('WHO Region')['Confirmed'].agg([
    'mean',
    'median',
    lambda x: x.mode()[0]
])

group_summary.columns = ['Mean Confirmed', 'Median Confirmed', 'Mode Confirmed']
print(group_summary)
```

```python

mean_val = df['Confirmed'].mean()
median_val = df['Confirmed'].median()

print('Skewness:', df['Confirmed'].skew())

if mean_val > median_val:
    print('Distribution is right (positively) skewed')
elif mean_val < median_val:
    print('Distribution is left (negatively) skewed')
else:
    print('Distribution is approximately symmetric')
```
