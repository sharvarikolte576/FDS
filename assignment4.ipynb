```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\sharv\OneDrive\Desktop\ML\loan_data.csv")
print(df.head())
```

```python
print(df.info())
```

```python
data_range = df['installment'].max() - df['installment'].min()
print(data_range)
```

```python
mean_value = df['installment'].mean()
print(mean_value)
```

```python
variance = df['installment'].var()
print(variance)
```

```python
std_deviation = df['installment'].std()
print(std_deviation)
```

```python
reg_coef = (std_deviation / mean_value) * 100
print(reg_coef)
```
